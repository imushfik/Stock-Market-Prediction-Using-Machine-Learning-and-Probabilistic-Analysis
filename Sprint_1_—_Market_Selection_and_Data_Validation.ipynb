```python
import sys
import pandas as pd

from fastapi.testclient import TestClient

# Locate your existing Sprint 1 backend
backend_path = "/content/sprint1_backend/sprint1_backend"

if backend_path not in sys.path:
    sys.path.insert(0, backend_path)

# Import existing FastAPI application
from main import app

client = TestClient(app)

# Retrieve historical stock data
response = client.get(
    "/api/stock/historical",
    params={
        "symbol": "AAPL",
        "start_date": "2022-01-01",
        "end_date": "2026-09-30"
    }
)

print("HTTP Status:", response.status_code)

if response.status_code == 200:
    result = response.json()

    # Convert stock data into a DataFrame
    stock_df = pd.DataFrame(result["data"])

    print("\nTotal rows:", len(stock_df))
    print("\nColumns:", stock_df.columns.tolist())

    display(stock_df.head())

else:
    print("Error:", response.text)
```

```python
import pandas as pd
import numpy as np

# Create a copy of the original stock data
df = stock_df.copy()

# Convert Date into a proper datetime format
df["Date"] = pd.to_datetime(df["Date"], utc=True)

# Sort from oldest to newest
df = df.sort_values("Date").reset_index(drop=True)

# Ensure price and volume columns are numeric
numeric_cols = ["Open", "High", "Low", "Close", "Volume"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Check missing values
print("Missing values BEFORE cleaning:")
print(df.isnull().sum())

# Remove duplicate dates
df = df.drop_duplicates(subset=["Date"])

# Remove rows with missing required values
df = df.dropna(subset=numeric_cols)

# Remove invalid prices and trading volume
valid_prices = (df[["Open", "High", "Low", "Close"]] > 0).all(axis=1)
valid_volume = df["Volume"] >= 0

df = df.loc[valid_prices & valid_volume].copy()

# Sort again and reset index
df = df.sort_values("Date").reset_index(drop=True)

print("\nMissing values AFTER cleaning:")
print(df.isnull().sum())

print("\nDataset shape:", df.shape)

print("\nFirst five rows:")
display(df.head())

print("\nLast five rows:")
display(df.tail())
```

```python
import pandas as pd
import numpy as np

# Copy the cleaned dataset
ml_df = df.copy()

# 1. Daily percentage return
ml_df["Return_1d"] = ml_df["Close"].pct_change()

# 2. Five-day percentage return
ml_df["Return_5d"] = ml_df["Close"].pct_change(5)

# 3. Moving averages
ml_df["MA_5"] = ml_df["Close"].rolling(5).mean()
ml_df["MA_10"] = ml_df["Close"].rolling(10).mean()
ml_df["MA_20"] = ml_df["Close"].rolling(20).mean()

# 4. Ten-day volatility
ml_df["Volatility_10"] = (
    ml_df["Return_1d"].rolling(10).std()
)

# 5. Trading volume ratio
ml_df["Volume_MA_10"] = (
    ml_df["Volume"].rolling(10).mean()
)

ml_df["Volume_Ratio"] = (
    ml_df["Volume"] /
    ml_df["Volume_MA_10"].replace(0, np.nan)
)

# 6. Price compared with 20-day moving average
ml_df["Price_vs_MA20"] = (
    ml_df["Close"] / ml_df["MA_20"] - 1
)

# 7. Daily price range
ml_df["Daily_Range"] = (
    (ml_df["High"] - ml_df["Low"]) /
    ml_df["Close"]
)

# 8. Next trading day's closing price
ml_df["Next_Close"] = ml_df["Close"].shift(-1)

# 9. Target: next-day stock direction
ml_df["Target"] = np.where(
    ml_df["Next_Close"].notna(),
    (ml_df["Next_Close"] > ml_df["Close"]).astype(int),
    np.nan
)

# List of features for machine learning
feature_cols = [
    "Return_1d",
    "Return_5d",
    "MA_5",
    "MA_10",
    "MA_20",
    "Volatility_10",
    "Volume_Ratio",
    "Price_vs_MA20",
    "Daily_Range"
]

# Remove rows without complete features or target
ml_df = ml_df.replace([np.inf, -np.inf], np.nan)
ml_df = ml_df.dropna(
    subset=feature_cols + ["Target"]
).reset_index(drop=True)

ml_df["Target"] = ml_df["Target"].astype(int)

print("Feature Engineering Completed!")

print("\nOriginal rows:", len(df))
print("Rows after feature engineering:", len(ml_df))

print("\nNumber of features:", len(feature_cols))
print("Features:", feature_cols)

print("\nTarget distribution:")
print(ml_df["Target"].value_counts())

print("\nFirst five rows:")
display(
    ml_df[["Date", "Close"] + feature_cols + ["Target"]].head()
)
```

```python
import numpy as np
import pandas as pd

# Keep records in chronological order
ml_df = ml_df.sort_values("Date").reset_index(drop=True)

# Features and prediction target
X = ml_df[feature_cols].copy()
y = ml_df["Target"].copy()

# 80% training and 20% testing
split_index = int(len(ml_df) * 0.80)

# Remove one training observation at the boundary.
# Its next-day target would use the first test day's price.
X_train = X.iloc[:split_index - 1].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index - 1].copy()
y_test = y.iloc[split_index:].copy()

print("Chronological Train/Test Split Completed!")

print("\nTotal usable records:", len(ml_df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nNumber of features:", X_train.shape[1])

print("\nTraining period:")
print("Start:", ml_df.loc[X_train.index[0], "Date"])
print("End:", ml_df.loc[X_train.index[-1], "Date"])

print("\nTesting period:")
print("Start:", ml_df.loc[X_test.index[0], "Date"])
print("End:", ml_df.loc[X_test.index[-1], "Date"])

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

# Verify that data is correctly separated
assert X_train.index.max() < X_test.index.min()

assert X_train.notna().all().all()
assert X_test.notna().all().all()

print("\nAll checks passed!")
```

```python
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# 1. Define models
models = {
    "Majority Baseline": DummyClassifier(
        strategy="most_frequent"
    ),

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=5,
        min_samples_leaf=10,
        random_state=42,
        n_jobs=-1
    )
}

# 2. Train and evaluate models
results = []

trained_models = {}

for name, model in models.items():

    print("\n" + "=" * 50)
    print("Training:", name)
    print("=" * 50)

    # Train using historical training data only
    model.fit(X_train, y_train)

    trained_models[name] = model

    # Predictions
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    # Performance metrics
    train_accuracy = accuracy_score(
        y_train, train_pred
    )

    test_accuracy = accuracy_score(
        y_test, test_pred
    )

    precision = precision_score(
        y_test, test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test, test_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test, test_pred,
        zero_division=0
    )

    # Store results
    results.append({
        "Model": name,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    print("Training Accuracy:", round(train_accuracy, 4))
    print("Testing Accuracy :", round(test_accuracy, 4))
    print("Precision        :", round(precision, 4))
    print("Recall           :", round(recall, 4))
    print("F1 Score         :", round(f1, 4))

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        test_pred,
        zero_division=0
    ))

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, test_pred))

# 3. Create comparison table
results_df = pd.DataFrame(results)

print("\n" + "=" * 55)
print("FINAL MODEL COMPARISON")
print("=" * 55)

display(
    results_df.sort_values(
        "F1 Score",
        ascending=False
    ).round(4)
)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score
)

# Step 1: Time-series cross-validation
tscv = TimeSeriesSplit(n_splits=4, gap=1)

# Step 2: Define candidate models
candidate_models = {
    "Majority Baseline": DummyClassifier(
        strategy="most_frequent"
    )
}

# Logistic Regression configurations
for c in [0.01, 0.1, 1]:
    candidate_models[f"Logistic C={c}"] = Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            C=c,
            max_iter=2000,
            random_state=42
        ))
    ])

# Random Forest configurations
for depth in [2, 3, 4]:
    candidate_models[f"RF Depth={depth}"] = (
        RandomForestClassifier(
            n_estimators=150,
            max_depth=depth,
            min_samples_leaf=20,
            random_state=42,
            n_jobs=-1
        )
    )

# Step 3: Evaluate each model
cv_results = []

for name, model in candidate_models.items():

    accuracies = []
    balanced_accuracies = []
    f1_scores = []

    for train_idx, val_idx in tscv.split(X_train):

        X_tr = X_train.iloc[train_idx]
        X_val = X_train.iloc[val_idx]

        y_tr = y_train.iloc[train_idx]
        y_val = y_train.iloc[val_idx]

        model.fit(X_tr, y_tr)

        predictions = model.predict(X_val)

        accuracies.append(
            accuracy_score(y_val, predictions)
        )

        balanced_accuracies.append(
            balanced_accuracy_score(y_val, predictions)
        )

        f1_scores.append(
            f1_score(y_val, predictions, zero_division=0)
        )

    cv_results.append({
        "Model": name,
        "CV Accuracy": np.mean(accuracies),
        "CV Balanced Accuracy": np.mean(balanced_accuracies),
        "CV F1": np.mean(f1_scores)
    })

# Step 4: Display results
cv_results_df = pd.DataFrame(cv_results)

cv_results_df = cv_results_df.sort_values(
    "CV Balanced Accuracy",
    ascending=False
).reset_index(drop=True)

print("=" * 55)
print("TIME-SERIES CROSS-VALIDATION RESULTS")
print("=" * 55)

display(cv_results_df.round(4))

print("\nBest validation model:")
print(cv_results_df.loc[0, "Model"])

print("\nStep 8A Completed Successfully!")
```

```python
from pathlib import Path
import shutil

backend_dir = Path(
    "/content/sprint1_backend/sprint1_backend"
)

main_file = backend_dir / "main.py"
model_file = backend_dir / "sprint2_stock_model.joblib"

# Check required files
assert main_file.exists(), "main.py not found"
assert model_file.exists(), "Saved ML model not found"

# Identify Sprint 2 code
marker = "# ===== SPRINT 2 ML PREDICTION ====="

# Code to append to main.py
sprint2_code = r'''

# ===== SPRINT 2 ML PREDICTION =====

import joblib
import numpy as np

from pathlib import Path
from datetime import timezone, timedelta
from functools import lru_cache


class StockPredictionRequest(BaseModel):
    symbol: str


@lru_cache(maxsize=1)
def load_prediction_bundle():
    model_path = (
        Path(__file__).resolve().parent
        / "sprint2_stock_model.joblib"
    )

    if not model_path.exists():
        raise HTTPException(
            status_code=503,
            detail="Machine learning model not found."
        )

    return joblib.load(model_path)


def prepare_prediction_features(df):

    data = df.copy()

    required_cols = [
        "Open", "High", "Low", "Close", "Volume"
    ]

    # Sort historical data
    data = data.sort_index()

    # Convert values to numeric
    for col in required_cols:
        data[col] = pd.to_numeric(
            data[col], errors="coerce"
        )

    # Clean historical records
    data = data.dropna(subset=required_cols)

    valid_prices = (
        data[["Open", "High", "Low", "Close"]] > 0
    ).all(axis=1)

    data = data.loc[
        valid_prices & (data["Volume"] >= 0)
    ].copy()

    if len(data) < 30:
        raise HTTPException(
            status_code=422,
            detail="Insufficient historical stock data."
        )

    # Recreate Sprint 2 training features
    data["Return_1d"] = data["Close"].pct_change()

    data["Return_5d"] = data["Close"].pct_change(5)

    data["MA_5"] = (
        data["Close"].rolling(5).mean()
    )

    data["MA_10"] = (
        data["Close"].rolling(10).mean()
    )

    data["MA_20"] = (
        data["Close"].rolling(20).mean()
    )

    data["Volatility_10"] = (
        data["Return_1d"].rolling(10).std()
    )

    data["Volume_MA_10"] = (
        data["Volume"].rolling(10).mean()
    )

    data["Volume_Ratio"] = (
        data["Volume"] /
        data["Volume_MA_10"].replace(0, np.nan)
    )

    data["Price_vs_MA20"] = (
        data["Close"] / data["MA_20"] - 1
    )

    data["Daily_Range"] = (
        (data["High"] - data["Low"]) /
        data["Close"]
    )

    bundle = load_prediction_bundle()
    features = bundle["features"]

    latest = data[features].tail(1)

    latest = latest.replace(
        [np.inf, -np.inf], np.nan
    )

    if latest.isna().any().any():
        raise HTTPException(
            status_code=422,
            detail="Latest stock features contain invalid values."
        )

    return latest, data


@app.post("/api/stock/predict")
def predict_stock(request: StockPredictionRequest):

    symbol = request.symbol.strip().upper()

    if not symbol:
        raise HTTPException(
            status_code=400,
            detail="Stock symbol cannot be empty."
        )

    # Load saved Sprint 2 model
    bundle = load_prediction_bundle()

    # Model currently supports AAPL only
    if symbol != bundle["symbol"]:
        raise HTTPException(
            status_code=400,
            detail=(
                "This model currently supports "
                f"{bundle['symbol']} only."
            )
        )

    # Retrieve approximately 120 days of data
    # End date is exclusive, avoiding the
    # current potentially incomplete trading day.
    end_date = datetime.now(timezone.utc).date()

    start_date = end_date - timedelta(days=120)

    try:
        historical = fetch_historical_data(
            symbol,
            str(start_date),
            str(end_date)
        )
    except Exception as exc:
        raise HTTPException(
            status_code=502,
            detail="Market data retrieval failed."
        ) from exc

    if historical.empty:
        raise HTTPException(
            status_code=404,
            detail="No historical stock data available."
        )

    # Calculate technical indicators
    features, cleaned_data = (
        prepare_prediction_features(historical)
    )

    # Predict using saved model
    model = bundle["model"]

    prediction = int(model.predict(features)[0])

    probabilities = model.predict_proba(features)[0]

    classes = list(model.classes_)

    probability_up = float(
        probabilities[classes.index(1)]
    )

    direction = (
        "UP" if prediction == 1
        else "DOWN / NOT UP"
    )

    # Return prediction results
    return {
        "symbol": symbol,
        "model": bundle["model_name"],
        "prediction": prediction,
        "direction": direction,
        "probability_up": round(probability_up, 4),
        "probability_down": round(
            1 - probability_up, 4
        ),
        "last_data_date": str(
            cleaned_data.index[-1].date()
        ),
        "historical_rows": len(cleaned_data),
        "status": "success",
        "disclaimer": (
            "Experimental machine learning prediction. "
            "Not financial advice."
        )
    }
'''

# Append Sprint 2 code only once
current_code = main_file.read_text(encoding="utf-8")

if marker not in current_code:

    backup = backend_dir / "main_sprint1_backup.py"

    shutil.copy2(main_file, backup)

    with open(main_file, "a", encoding="utf-8") as f:
        f.write("\n\n" + sprint2_code)

    print("Sprint 2 ML endpoint added successfully!")
    print("Backup created:", backup)

else:
    print("Sprint 2 endpoint already exists.")

# Check Python syntax
import ast

updated_code = main_file.read_text(encoding="utf-8")
ast.parse(updated_code)

print("\nPython syntax check: PASSED")
print("Updated main.py:", main_file)
print("Total lines:", len(updated_code.splitlines()))

print("\nStep 11B completed!")
```

```python
import subprocess
import sys
import time
import requests
from pathlib import Path

backend_dir = Path(
    "/content/sprint1_backend/sprint1_backend"
)

port = 8001

url = f"http://127.0.0.1:{port}"

# Check whether server is already running
try:
    response = requests.get(
        url + "/openapi.json",
        timeout=2
    )
    server_running = response.status_code == 200
except requests.RequestException:
    server_running = False

if not server_running:

    log_path = backend_dir / "sprint2_server.log"

    log_file = open(log_path, "a")

    server_process = subprocess.Popen(
        [
            sys.executable,
            "-m", "uvicorn",
            "main:app",
            "--host", "0.0.0.0",
            "--port", str(port)
        ],
        cwd=str(backend_dir),
        stdout=log_file,
        stderr=subprocess.STDOUT
    )

    print("Starting Sprint 2 FastAPI server...")

    for attempt in range(15):
        time.sleep(1)

        try:
            response = requests.get(
                url + "/openapi.json",
                timeout=2
            )

            if response.status_code == 200:
                server_running = True
                break

        except requests.RequestException:
            pass

print("Server running:", server_running)

if server_running:
    response = requests.get(
        url + "/demo-sprint2",
        timeout=10
    )

    print("HTML Status:", response.status_code)
    print("Sprint 2 server is ready!")
else:
    print("Server failed to start.")
    print("Check sprint2_server.log")
```
